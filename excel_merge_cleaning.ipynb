# Excel Merge Automation with Python & Pandas

This project demonstrates how multiple Excel files can be combined into one dataset using Python and Pandas.

## 1. Import Libraries and Load Excel Files

Import Pandas and load monthly sales datasets for analysis.

```python
import pandas as pd

jan = pd.read_excel("sales_jan.xlsx")

feb = pd.read_excel("sales_Feb.xlsx")

mar = pd.read_excel("sales_Mar.xlsx")
```

## 2. Preview Individual Datasets

Inspect the first rows of each dataset before merging.

```python
print(jan.head())

print(feb.head())

print(mar.head())
```

## 3. Merge Multiple Excel Files

Combine monthly sales data into one dataset using pd.concat().

```python
merged = pd.concat(
    [jan, feb, mar],
    ignore_index=True
)
print(merged)
```

## 4. Inspect Merged Dataset

```python
print(merged.info())

print(merged.isnull().sum())

print(merged.duplicated().sum())
```

## 5. Remove Duplicate Records

```python
merged = merged.drop_duplicates()
```

# 6. Convert Invalid Data Types

```python
merged["Sales"] = pd.to_numeric(
    merged["Sales"],
    errors="coerce"
)
```

## 7. Handle Missing Values

```python


merged["Sales"] = merged["Sales"].fillna(
    merged["Sales"].mean()
)

merged["Month"] = merged["Month"].fillna(
    "Unknown"
)
```

## 8. Validate Merged Dataset

Check data types, row counts and summary statistics after merging.

```python
print(merged.info())

print(merged.describe().round(2))
```

## 9. Export Final Dataset

Save the merged dataset as a new Excel file.

```python
merged.to_excel(
    "merged_sales.xlsx",
    index=False
)
```

# Project Outcome

This project successfully merged multiple Excel files, handled missing values, removed duplicate records, and exported a cleaned dataset using Python and Pandas.
